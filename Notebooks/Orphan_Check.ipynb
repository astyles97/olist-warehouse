```python
import os
from dotenv import load_dotenv
import pandas as pd
from sqlalchemy import create_engine
import numpy as np
import matplotlib.pyplot as plt

load_dotenv()

engine = create_engine(os.environ['DATABASE'])
```

```python
orphan_sellers = pd.read_sql("SELECT seller_id "
                             "FROM raw.sellers "
                             "WHERE seller_id NOT IN (SELECT DISTINCT seller_id FROM raw.order_items)",
                             engine)
```

```python
no_reviews = pd.read_sql("SELECT order_id "
                        "FROM raw.order_items "
                        "WHERE order_id NOT IN (SELECT DISTINCT order_id FROM raw.reviews)", engine)
```

```python
print(f"Number of orphan sellers: {len(orphan_sellers)}")
print(f"Number of orders without reviews: {len(no_reviews)}")
# 
# Number of orphan sellers: 0
# Number of orders without reviews: 942
```
